Furniture sales went from:

$1,371.48 → $11,252.11

That's roughly an 8.2× increase.

And June's total sales were $18,068.09, meaning furniture alone contributed about 62% of June's sales.

So we've discovered a much stronger explanation for June's high AOV:

June's unusually high sales were primarily driven by a dramatic increase in Furniture sales, rather than simply a broad increase across all product categories.

```python
feb_furniture = feb_2025[feb_2025["Product_Category"] == "Furniture"]
jun_furniture = jun_2025[jun_2025["Product_Category"]== "Furniture"]
```

```python
print("February furniture quantity:", feb_furniture["Quantity"].sum())
print("June furniture quantity:", jun_furniture["Quantity"].sum())
```

```python
print("feb furniture order",len(feb_furniture))
print("feb furniture order",len(jun_furniture))
```

```python
print("February furniture Aov:", feb_furniture["Total_Sales"].mean())
print("June furniture Aov:", jun_furniture["Total_Sales"].mean())
```

June 2025's sales spike was primarily driven by Furniture. Furniture revenue increased approximately 720% versus February, supported by a 150% increase in orders, a 240% increase in units sold, and a 228% increase in average order value.
