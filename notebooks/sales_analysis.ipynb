```python
region_analysis = (
    df.groupby("Region")
    .agg(
        Orders=("Order_ID", "count"),
        Sales=("Total_Sales", "sum"),
        Profit=("Profit", "sum")
    )
)

region_analysis
```

```python
region_analysis["Profit_Margin"] = (
    region_analysis["Profit"] /
    region_analysis["Sales"]
) * 100
```

```python
region_analysis.sort_values("Sales", ascending=False)
```

```python
region_analysis["AOV"] = (
    region_analysis["Sales"] /
    region_analysis["Orders"]
)
```

```python
region_analysis.sort_values("AOV", ascending=False)
```

```python
region_analysis["Sales"].sort_values(ascending=False).plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.show()
```

```python
region_analysis["Profit_Margin"].sort_values(ascending=False).plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Profit Margin by Region")
plt.xlabel("Region")
plt.ylabel("Profit Margin (%)")
plt.xticks(rotation=45)

plt.show()
```

```python
country_analysis = (
    df.groupby("Country")
    .agg(
        Orders=("Order_ID", "count"),
        Sales=("Total_Sales", "sum"),
        Profit=("Profit", "sum")
    )
)

country_analysis["Profit_Margin"] = (
    country_analysis["Profit"] /
    country_analysis["Sales"]
) * 100

country_analysis["AOV"] = (
    country_analysis["Sales"] /
    country_analysis["Orders"]
)

country_analysis.sort_values("Sales", ascending=False)
```
